# 🧠 Natural Language Processing: Rotten Tomatoes Critic Review Sentiment Engine

This notebook trains a Natural Language Processing (NLP) sentiment classifier on over 54,000 Rotten Tomatoes critic reviews using TF-IDF vectorization and Logistic Regression to classify reviews as 'Fresh' or 'Rotten'.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score
```

## 1. Load Review Data & Text Preprocessing

```python
rt_reviews = pd.read_csv('Data/rt.reviews.tsv', sep='\t', encoding='latin1').dropna(subset=['review', 'fresh'])
rt_reviews['label'] = (rt_reviews['fresh'] == 'fresh').astype(int)

sample_df = rt_reviews.sample(n=min(15000, len(rt_reviews)), random_state=42)

tfidf = TfidfVectorizer(max_features=2500, stop_words='english', ngram_range=(1, 2))
X_nlp = tfidf.fit_transform(sample_df['review'])
y_nlp = sample_df['label']
```

## 2. Train Sentiment Classifier & Evaluation

```python
X_train_nlp, X_test_nlp, y_train_nlp, y_test_nlp = train_test_split(X_nlp, y_nlp, test_size=0.2, random_state=42)
clf = LogisticRegression(max_iter=1000)
clf.fit(X_train_nlp, y_train_nlp)

y_pred_nlp = clf.predict(X_test_nlp)
print(f'Logistic Regression Accuracy: {clf.score(X_test_nlp, y_test_nlp):.4f}')
print(f'ROC-AUC Score: {roc_auc_score(y_test_nlp, clf.predict_proba(X_test_nlp)[:, 1]):.4f}')
```

## 3. Confusion Matrix Visualization

```python
cm = confusion_matrix(y_test_nlp, y_pred_nlp)
plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Rotten', 'Fresh'], yticklabels=['Rotten', 'Fresh'])
plt.title('Rotten Tomatoes Critic Review Sentiment Confusion Matrix', fontsize=12, fontweight='bold')
plt.xlabel('Predicted Sentiment')
plt.ylabel('Actual Sentiment')
plt.tight_layout()
plt.show()
```
